```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


train = pd.read_csv('train.csv')

train.head()
```

```python
train.info()
```

```python
train.describe()
```

```python
train.isnull().sum()
```

```python
train.isnull()
```

```python
train['Embarked'].fillna(train['Embarked'].mode()[0], inplace=True)
```

```python
train.isnull().sum()
```

```python
train['Cabin'].fillna('Unknown', inplace=True)
train.isnull().sum()
```

```python
train['Age'].fillna(train['Age'].median(), inplace=True)
train.isnull().sum()
```

```python
# Plot histogram for Age and Fare
train['Age'].hist(bins=30, edgecolor='black', color='skyblue', alpha=0.7)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

train['Fare'].hist(bins=30, edgecolor='black', color='skyblue', alpha=0.7)
plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()
```

```python
# Count plot for 'Survived' (target variable)
sns.countplot(x='Survived', data=train, palette='Set2')
plt.title('Survival Count')
plt.show()

# Count plot for 'Sex'
sns.countplot(x='Sex', data=train, palette='Set2')
plt.title('Gender Count')
plt.show()

# Count plot for 'Embarked'
sns.countplot(x='Embarked', data=train, palette='Set2')
plt.title('Embarkation Port Count')
plt.show()
```

```python
sns.boxplot(x='Survived', y='Age', data=train, palette='Set2')
plt.title('Age Distribution by Survival')
plt.show()
```

```python
sns.boxplot(x='Survived', y='Fare', data=train, palette='Set2')
plt.title('Fare Distribution by Survival')
plt.show()
```

```python
# Calculate the correlation matrix
corr = train[['Age', 'Fare', 'SibSp', 'Parch']].corr()

# Create a heatmap
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```

```python
sns.pairplot(train[['Age', 'Fare', 'SibSp', 'Parch', 'Survived']], hue='Survived', palette='Set2')
plt.show()
```

```python
sns.barplot(x='Sex', y='Survived', data=train, palette='Set2')
plt.title('Survival Rate by Gender')
plt.show()
```

```python
import numpy as np

# Check skewness of 'Fare'
print(train['Fare'].skew())

# Apply log transformation to 'Fare' if it's highly skewed
train['Fare'] = train['Fare'].apply(lambda x: np.log(x+1) if x > 0 else 0)
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

# Create the feature matrix
X = train[['Age', 'Fare', 'SibSp', 'Parch']]

# Add a constant to the model
X = add_constant(X)

# Calculate VIF for each feature
vif = pd.DataFrame()
vif['Feature'] = X.columns
vif['VIF'] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

print(vif)

```
